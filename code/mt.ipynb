```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
```

```python
import os
import glob
fl = sorted(glob.glob("../dataset/translated/nllb/babylm_data_ind/babylm_*/*"))
os.path.splitext(fl[0])[0]
```

```python
fl[0]
```

```python
tokenizer = AutoTokenizer.from_pretrained("../models/opus-mt-en-id")
model = AutoModelForSeq2SeqLM.from_pretrained("../models/opus-mt-en-id")
```

```python
with open('../dataset/babylm_data/babylm_10M/aochildes.train', 'r') as f:
    aochildes = f.readlines()
aochildes[:20]
```

```python
inputs = tokenizer(aochildes[:20], return_tensors="pt", padding=True)
translated_tokens = model.generate(**inputs, max_length=512)
translated_text = tokenizer.decode(translated_tokens[0], skip_special_tokens=True)

print(translated_text)
```

```python
for token in translated_tokens:
    translated_text = tokenizer.decode(token, skip_special_tokens=True)
    print(translated_text)
```
